## Feature engineering

```python
df_fintech['transaction_id'] = range(1,len(df_fintech)+1)
```

```python
df_fintech['transaction_day'] = ((df_fintech['step']-1)//24)+1
```

```python
df_fintech['hour_of_day'] = (df_fintech['step'] - 1) % 24
```

```python
df_fintech['transaction_amount_band'],bins = pd.qcut(
    df_fintech['amount'],
    q=4,
    labels=['Low', 'Medium', 'High','Very High'],
    retbins=True
)
for i, label in enumerate(['Low', 'Medium', 'High','Very High']):
    print(f"{label}: {bins[i]} - {bins[i+1]}")
```

```python
high_value_threshold = df_fintech['amount'].quantile(0.95)
df_fintech['is_high_value'] = (df_fintech['amount']>high_value_threshold).astype(int)
```

```python
df_fintech["is_origin_balance_drained"] = (
    (df_fintech["origin_new_balance"] == 0) &
    (df_fintech["origin_old_balance"] > 0)
).astype(int)
```
